```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("StudentsPerformance.csv")
df.head()
```

```python
df.drop_duplicates(inplace=True)

df.fillna(df.mean(numeric_only=True), inplace=True)

print("Dataset Shape:", df.shape)
print("Missing Values:\n", df.isnull().sum())
```

```python
math_scores = np.array(df["math score"])

print("Mean Math Score:", np.mean(math_scores))
print("Standard Deviation:", np.std(math_scores))
print("Max Math Score:", np.max(math_scores))
print("Min Math Score:", np.min(math_scores))
```

```python
plt.figure(figsize=(7, 5))

plt.hist(df["math score"], bins=15, color="skyblue", edgecolor="black")

plt.title("Math Score Distribution")
plt.xlabel("Math Score")
plt.ylabel("Frequency")

plt.show()
```

The math scores follow a near-normal bell-curve distribution centered around 60–70. Most students score between 50 and 80, with very few scoring below 30 or achieving a perfect 100.

```python
plt.figure(figsize=(7, 5))

plt.scatter(df["math score"], df["reading score"], alpha=0.6, color="teal")

plt.title("Math Score vs Reading Score")
plt.xlabel("Math Score")
plt.ylabel("Reading Score")

plt.show()
```

There is a strong positive linear relationship between math scores and reading scores. Students who score higher in reading consistently perform better in mathematics.

```python
ethnicity_math = df.groupby("race/ethnicity")["math score"].mean()

plt.figure(figsize=(8, 5))

ethnicity_math.plot(kind="bar", color="coral", edgecolor="black")

plt.title("Average Math Score by Race/Ethnicity")
plt.xlabel("Race/Ethnicity Group")
plt.ylabel("Average Math Score")
plt.xticks(rotation=0)

plt.show()
```

Group E demonstrates the highest average math score among all demographic groups, whereas Group A reflects the lowest mean score.

```python
sorted_reading = df["reading score"].sort_values().reset_index(drop=True)

plt.figure(figsize=(7, 5))

plt.plot(sorted_reading, color="purple")

plt.title("Sorted Reading Scores Trend")
plt.xlabel("Student Index (Sorted)")
plt.ylabel("Reading Score")

plt.show()
```

The sorted reading score plot illustrates the cumulative progression of performance across all students, rising steadily from the lowest score to the maximum of 100.

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].hist(
    df["writing score"], bins=15, color="mediumseagreen", edgecolor="black"
)
axes[0].set_title("Writing Score Distribution")
axes[0].set_xlabel("Writing Score")
axes[0].set_ylabel("Frequency")

axes[1].scatter(df["reading score"], df["writing score"], alpha=0.6, color="orchid")
axes[1].set_title("Reading vs Writing Score")
axes[1].set_xlabel("Reading Score")
axes[1].set_ylabel("Writing Score")

plt.tight_layout()
plt.show()
```
