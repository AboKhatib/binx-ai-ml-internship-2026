## 1. Variables and Basic Data Types

Python variables store values. Common data types include integers, floats, strings, and booleans.

```python
name = "Ahmad"
age = 22
gpa = 4
is_student = True

print(name)
print(age)
print(gpa)
print(is_student)
```

```python
print(type(name))
print(type(age))
print(type(gpa))
print(type(is_student))
```

## 2. Collection Data Types

Python provides several collection types for storing multiple values:
- Lists
- Tuples
- Dictionaries
- Sets

```python
courses = ["Python", "Data scince", "Machine Learning"]
location = ("Nablus", "Palestine")
student = {
    "name" : "Ahmad",
    "age" : 22,
    "gpa" : 4
}
skills = {"Python", "Git", "Python", "Jupyter"}

print(courses)
print(location)
print(student)
print(skills)
```

```python
print(type(courses))
print(type(location))
print(type(student))
print(type(skills))
```

## 3. Conditional Statements

Conditional statements allow a program to make decisions based on whether a condition is true or false.

```python
score = 78

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
elif score >= 60:
    print("Grade: D")
else:
    print("Grade: F")
```

```python
number = -5

if number >0:
    print("Positive")
elif number <0:
    print("Negative")
else:
    print("Zero")
```

## 4. For Loops

A `for` loop repeats a block of code for each item in a sequence.

```python
courses = ["Python","Data Scince","Machine Learning"]

for course in courses:
    print(course)
```

```python
for number in range (1,6):
    print (number)
```

```python
for number in range (1,21):
    if number%2 == 0:
        print(number)
```

## 5. While Loops

A `while` loop repeats code as long as a condition remains true.

```python
count = 1

while count<=5:
    print (count)
    count+=1
```

```python
number = 2

while number <=10:
    print (number)
    number+=2
```

## 6. Functions

Functions group reusable code into a named block. They can receive inputs called parameters and return a result.

```python
def greet(name):
    print(f"Hello, {name}!")

greet("Ahmad")
```

```python
def square (number):
    return number ** 2

square(3)
```

```python
def average (numbers):
    return sum(numbers)/len(numbers)

values = [10,20,30,40]
average (values)
```

```python
def maximum(numbers):
    return max(numbers)

values = [8, 15, 3, 22, 10]
print(maximum(values))
```

## 7. List Comprehensions

A list comprehension creates a new list using a compact form of a `for` loop.

```python
squares = []

for number in range(1, 6):
    squares.append(number ** 2)

print(squares)
```

```python
squares = [number ** 2 for number in range(1, 6)]

print(squares)
```

```python
even_numbers = [number for number in range(1, 21) if number % 2 == 0]

print(even_numbers)
```

```python
even_squares = [number ** 2 for number in range(1, 11) if number % 2 == 0]

print(even_squares)
```

## 8. Object-Oriented Programming Basics

A class is a blueprint for creating objects. Objects can store data in attributes and perform actions through methods.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"My name is {self.name}, and I am {self.age} years old.")

student1 = Student("Ahmad", 22)

student1.introduce()
```

```python
student2 = Student("Sara", 21)

student2.introduce()
```
