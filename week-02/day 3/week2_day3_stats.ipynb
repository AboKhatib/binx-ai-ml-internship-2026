# Linear Algebra for Machine Learning

## Why Linear Algebra Is the Language of Machine Learning

Every dataset in Machine Learning is represented as a matrix.

- Rows represent samples.
- Columns represent features.

Machine Learning models store their parameters as vectors or matrices, and training consists of mathematical operations on these objects.

Understanding vectors, matrices, dot products, and matrix multiplication is essential because these are the core operations behind every ML model.

import numpy as np

## Vectors

A vector is an ordered collection of numbers.

In Machine Learning, one data sample is represented as a vector.

Each value inside the vector is called a feature.

```python
student = np.array([22, 3.8])

print(student)
```

```python
house = np.array([180, 4, 250000])

print(house)
```

## Vector Shape

The shape tells us how many elements the vector contains.

```python
print(student.shape)
print(house.shape)
```

## Matrices

A matrix is a collection of vectors.

Rows represent samples.

Columns represent features.

```python
students = np.array([
    [22, 3.8],
    [20, 3.5],
    [25, 3.9],
    [21, 3.2]
])

print(students)
```

## Matrix Shape

Matrix shape is written as:

(rows, columns)

```python
print(students.shape)
```

## Accessing Matrix Elements

```python
print(students[0])

print(students[:,0])

print(students[:,1])
```

## Dot Product

The dot product combines two vectors into a single number.

Machine Learning models use the dot product to calculate predictions.

```python
x = np.array([2, 4])

w = np.array([3, 5])

result = np.dot(x, w)

print(result)
```

```python
scores = np.array([80, 90, 100])

weights = np.array([0.2, 0.3, 0.5])

prediction = np.dot(scores, weights)

print(prediction)
```

## Matrix Multiplication

Matrices can be multiplied only when their dimensions are compatible.

(number of columns in the first matrix)

must equal

(number of rows in the second matrix)

```python
A = np.array([
    [1,2],
    [3,4]
])

B = np.array([
    [5,6],
    [7,8]
])

C = np.matmul(A,B)

print(C)
```

```python
print(A.shape)

print(B.shape)

print(C.shape)
```

```python
person = np.array([25, 180, 75])

print(person)
```

```python
people = np.array([
    [25,180,75],
    [30,175,80],
    [22,170,65]
])

print(people)
```

```python
x = np.array([1,2,3])

w = np.array([4,5,6])

print(np.dot(x,w))
```

```python
A = np.array([
    [1,2],
    [3,4]
])

B = np.array([
    [2],
    [5]
])

C = np.matmul(A,B)

print(C)

print(C.shape)
```
