## 4. Mean (Average)

The mean is the arithmetic average of all values in a dataset.

It is calculated by adding all observations together and dividing by the total number of observations.

The mean provides a measure of the central tendency of the data and is commonly used when the data is approximately symmetric and does not contain extreme outliers.

```python
math_scores = df["Springsnails"]
```

```python
mean_score = np.mean(math_scores)

print("Mean:", mean_score)
```

## 5. Median

The median is the middle value in a sorted dataset.

If the dataset contains an odd number of observations, the median is the center value.

If the dataset contains an even number of observations, the median is the average of the two middle values.

Unlike the mean, the median is resistant to extreme values (outliers), making it a better measure of central tendency for skewed datasets.

```python
median_score = np.median(math_scores)

print("Median:", median_score)
```

```python
print("Mean  :", math_scores.mean())
print("Median:", math_scores.median())
```

## 6. Mode

The mode is the value that appears most frequently in a dataset.

Unlike the mean and median, the mode is determined by counting the frequency of each value.

The mode is useful for both numerical and categorical data, making it especially valuable when identifying the most common observation.

```python
mode_score = math_scores.mode()

print(mode_score)
```

## 7. Mean vs. Median 

Outliers are extreme values that differ significantly from the rest of the data.

The mean is sensitive to outliers because every value contributes to its calculation.

The median, however, depends only on the position of the values after sorting, making it much more robust when extreme values are present.

```python
import numpy as np

data = np.array([10, 12, 12, 13, 100])

print("Data:", data)
print("Mean:", np.mean(data))
print("Median:", np.median(data))
```
