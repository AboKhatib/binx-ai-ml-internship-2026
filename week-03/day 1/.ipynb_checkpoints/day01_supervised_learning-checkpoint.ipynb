# What is Machine Learning?

Machine Learning is a branch of Artificial Intelligence that allows computers to learn patterns from data and make predictions without being explicitly programmed.

# What is Supervised Learning?

Supervised Learning uses labeled data.

Each training example already contains the correct answer.

The model learns the relationship between the inputs and the correct output.

Examples:

Predict House Price

Predict Salary

Predict Student Grade

Spam Detection

Disease Prediction

Customer Churn Prediction

# Features (X)

Features are the information used by the model.

# Target (y)

Target is the value that we want to predict.

```python
df = pd.DataFrame({
    "sepal_length": [5.1, 4.9],
    "sepal_width": [3.5, 3.0],
    "species": ["setosa", "setosa"]
})

X = df.drop("species", axis=1)
```

```python
y = df["species"]
```

```python

X.head()
```

```python
y.head()
```

# Scikit-learn

Scikit-learn is the most popular Python Machine Learning library.

It provides a unified API for training and evaluating machine learning models.

The basic workflow is:

1. Import Model

2. Create Model

3. Train Model

4. Predict

5. Evaluate

Training Data

Used for learning.

Testing Data

Used only for evaluation.

The model should never see the test data during training.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
print(X_train.shape)
print(X_test.shape)

print(y_train.shape)
print(y_test.shape)
```

random_state=42 ensures that the train/test split is reproducible.

Every time the notebook runs, the same rows will be assigned to the training and testing sets.

test_size=0.20 means

80% Training Data

20% Testing Data
