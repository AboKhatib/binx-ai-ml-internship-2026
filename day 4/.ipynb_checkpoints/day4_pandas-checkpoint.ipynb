# Day 4 — Pandas: Tabular Data

## 1. Series and DataFrames

Pandas is the primary Python library for working with tabular, spreadsheet-like data.

A `Series` represents one labeled column.

A `DataFrame` represents a complete table containing rows and named columns.

## 2. Loading Data from a CSV File

Pandas uses `read_csv()` to load a CSV file into a DataFrame.

The CSV file should be in the same folder as the Jupyter Notebook, or its correct file path must be provided.

```python
import pandas as pd

df = pd.read_csv("BLNWR.csv")
```

## 3. Inspecting the Dataset

After loading the dataset, it is important to inspect its structure before performing any analysis.

This includes:
- Viewing the first few rows.
- Checking the number of rows and columns.
- Listing the column names.
- Identifying the data type of each column.

```python
print("First five rows:")
df.head()
```

```python
print("Shape:")
df.shape
```

```python
print("\nColumns:")
df.columns
```

```python
print("\nData types:")
df.columns.tolist()
```

```python
print("\nDatatypes:")
df.dtypes
```

```python
print("\nDataset information:")
df.info()
```

## 4. Checking for Missing Values

Before analyzing the data, the dataset should be checked for missing values.

Missing values can affect calculations and statistical analysis, so they should be identified before further processing.

```python
missing_values= df.isnull().sum()

print("Missing values per column: ")
print(missing_values)
```

```python
missing_percentage = (df.isnull().sum()/len(df)) *100

print("Missing-value percentage per column: ")
print(missing_percentage)
```

```python
average_noels = df["Noels_Alive"].mean()
print("Average Noels_Alive:", average_noels)
```

## 5. Filtering the Dataset

Filtering allows us to select only the rows that satisfy a specific condition.

For this analysis, the dataset is filtered to include observations where the value of `Noels_Alive` is greater than the average.

```python
filtered_df = df[df["Noels_Alive"] > average_noels]
filtered_df.head()
```

```python
print("Original number of rows:", df.shape[0])
print("Filtered number of rows:", filtered_df.shape[0])
filtered_df[["Noels_Alive", "Springsnails"]].describe()
```

```python
season_average = df.groupby("Season")["Noels_Alive"].mean()
print("Average Noels_Alive by season:")
print(season_average)
```

```python
season_summary = df.groupby("Season").agg({
    "Noels_Alive": "mean",
    "Springsnails": "mean"
})

print(season_summary)
```
