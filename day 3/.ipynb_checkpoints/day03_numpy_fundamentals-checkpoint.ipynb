```python
import numpy as np
print ("Numpy version: ", np.__version__)
```

```python
numbers = np.array([10, 20, 30, 40, 50])
print(numbers)
print(type(numbers))
```

## 2. Creating NumPy Arrays

NumPy provides several functions for creating arrays depending on the required data.

```python
array_from_list = np.array([1, 2, 3, 4, 5])
zeros_array = np.zeros(5)
ones_array = np.ones((2,3))
range_array = np.arange(0, 10, 2)
linear_array = np.linspace (0, 1, 5)

print("From list:", array_from_list)
print("Zeros:" ,zeros_array)
print("Ones:" ,ones_array)
print("Using range:" ,range_array)
print("Using linspace:" ,linear_array)
```

```python
sample_array = np.array([[10, 20, 30],[40, 50, 60]])

print("Array:\n", sample_array)
print("Shape\n", sample_array.shape)
print("Dimensions", sample_array.ndim)
print("Data type", sample_array.dtype)
print("Total elements", sample_array.size)
```

## 3. Indexing and Slicing

NumPy indexing selects individual elements, while slicing selects a range of elements.

```python
numbers = np.array([10,20,30,40,50])
print("first element:", numbers[0])
print("second element:", numbers[1])
print("last element:", numbers[-1])
```

```python
print("First three elements:", numbers[:3])
print("Elements from index 1 to 3:", numbers [1:4])
print("Every second element:", numbers [::2])
print("Reversed array:", numbers [::-1])
```

```python
matrix = np.array([
    [1,2,3],
    [4,5,6]
])

print("Matrix:\n", matrix)
print("Element at row 0,column 2", matrix [0,2])
print("Second column:", matrix [:,2])
print("Last row:", matrix [1,:])
```

## 4. Vectorized Operations

Vectorized operations apply calculations to an entire NumPy array without using an explicit Python loop.

```python
a = np.array([1, 2, 3, 4])

print("Original array", a)
print("Multiply by 2:", a * 2)
print("Add array to itself:", a+a)
print("Mean:", a.mean())
```

## 5. Boolean Masking

Boolean masking selects array values that satisfy a condition.

```python
a = np.array([1,2,3,4])

print("Original array:", a)
print("Condition:", a>2)
print("Values greater than 2:" , a[a>2])
```

## 6. Broadcasting

Broadcasting allows NumPy to combine arrays with different but compatible shapes.

```python
matrix = np.array([
    [1,2,3],
    [4,5,6]
])

row = np.array([10,20,30])

print(matrix + row)
```

# NumPy Array Manipulation Lab

```python
array_4x4 = np.arange (1,17).reshape(4,4)

print("Array:\n", array_4x4)
print("Shape:", array_4x4.shape)
print("Data type:", array_4x4.dtype)
```

```python
second_column = array_4x4[:,1]
last_row = array_4x4[-1,:]

print("Second column:", second_column)
print("Last row", last_row)
```

```python
array_mean = array_4x4.mean()
values_above_mean = array_4x4[array_4x4 > array_mean]

print("Mean:", array_mean)
print("Values greater than the mean:", values_above_mean)
```

```python
row = np.array([10, 20, 30, 40])
broadcast_result = array_4x4 + row

print("Row array:", row)
print("Broadcast result:\n", broadcast_result)
```
